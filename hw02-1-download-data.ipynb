# HW2.1: Download Hourly Wikipedia Event Data

Downloads the hourly `.parquet` Wikipedia event files from the public `s3://dsan6000-wikipedia` bucket into a local `data/` subfolder.

```python

import sys
print(sys.executable)
!{sys.executable} -m pip install boto3


import boto3
from botocore import UNSIGNED
from botocore.config import Config
import os
```

```python
# The source bucket is public, so we use an "unsigned" (anonymous) client
s3 = boto3.client("s3", config=Config(signature_version=UNSIGNED))

bucket = "dsan6000-wikipedia"
prefix = "hourly_parquet/"

os.makedirs("data", exist_ok=True)
```

```python
# List every object under that prefix
response = s3.list_objects_v2(Bucket=bucket, Prefix=prefix)
keys = [obj["Key"] for obj in response["Contents"]]
print(f"Found {len(keys)} files")
```

```python
# Download each one into ./data, keeping just the filename
for key in keys:
    filename = key.split("/")[-1]
    local_path = os.path.join("data", filename)
    s3.download_file(bucket, key, local_path)
    print(f"Downloaded {filename}")
```
