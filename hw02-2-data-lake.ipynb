# HW2.2: Storing OLTP Events in a Data Lake

Creates a new personal S3 bucket and uploads the locally-downloaded hourly `.parquet` files into a `wikipedia-hourly/` subfolder there.

```python
import boto3
import os
```

```python
netid = "kt917"
bucket_name = f"dsan6000-{netid}"
region = "us-east-1"  # match your AWS Academy lab's region

s3 = boto3.client("s3", region_name=region)
```

```python
# Create the bucket (us-east-1 doesn't need a LocationConstraint)
s3.create_bucket(Bucket=bucket_name)
print(f"Created bucket: {bucket_name}")
```

```python
# Upload every local parquet file into a wikipedia-hourly/ "subfolder"
local_dir = "data"
for filename in os.listdir(local_dir):
    if filename.endswith(".parquet"):
        local_path = os.path.join(local_dir, filename)
        s3_key = f"wikipedia-hourly/{filename}"
        s3.upload_file(local_path, bucket_name, s3_key)
        print(f"Uploaded {filename} to s3://{bucket_name}/{s3_key}")
```
